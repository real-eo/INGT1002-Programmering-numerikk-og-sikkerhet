# Øving 3

Hei og velkommen til tredje øving i Programmeringsdelen av emnet INGT1002 - Programmering, numerikk og sikkerhet. Hovedfokuset for denne øvingen er å lære seg bruk av `if`-setnigner og betingelser, samt bli kjent med løkker som et verktøy for å iterere seg gjennom en sekvens med verdier. 


## Læringsmål

- If-setninger, hvorfor og hvordan
- Betingelesr og logiske uttrykk
- Iterering gjennom sekvenser med bruk av for-løkke

## Godkjenning

Denne øvingen kan gi inntil 100 poeng. Kravet for godkjenning er 60 poeng eller mer.
