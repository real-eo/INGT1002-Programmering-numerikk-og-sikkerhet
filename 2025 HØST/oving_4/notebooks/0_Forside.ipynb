# Øving 4

Hei og velkommen til fjerde øving i Programmeringsdelen av emnet INGT1002 - Programmering, numerikk og sikkerhet. Hovedfokuset for denne øvingen er å lære mer om for-løkker, samt begynne å bruke while-løkker. I tillegg vil vi se på litt mer kompleks bruk av løkker og arrays med introduksjon av 2-dimensjonale arrays og nøstede løkker.


## Læringsmål

- For-løkker med bruk av `range` og indeksering
- While-løkker
- Arrays med 2 Dimensjoner
- Nøstede løkker

## Krav til godkjenning

Denne øvingen kan gi inntil 100 poeng. Kravet for godkjenning er 60 poeng eller mer. (Men gjør gjerne alt.)
